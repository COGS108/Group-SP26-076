## Research Question

Which factors predict NPA MVP voting share from 1996 to 2022, and how has their importance shifted across the eras?

- Which player statistics (points, assists, rebounds) and advanced efficiency metrics (BPM, Win Shares) are the strongest predictors of MVP vote share across the entire 1996-2022 period?
- How does the predictive importance of traditional versus advanced metrics differentiate between the pre-analytics era (1996-2004) and modern era (2004-2022)?

Our dataset will likely combine per-game player statistics from Kaggle using manually added MVP vote sharing data from Basketball Reference. This means this will be mainly a prediction + statistical interference task, and would require us to use regression-based models in order to count each statistic’s contribution to the vote share, and then compare the importance of features across both eras to identify whether the “MVP profile” changed alongside advanced analytics becoming mainstream.


## Background and Prior Work

The National Basketball Association (NBA) is the predominant professional basketball league in the United States. When people talk about the NBA, a common point of discussion is who the face of the league is. A tangible way to assign this title is by looking at which player is awarded the league MVP for each NBA season. The MVP award for the NBA was introduced in 1956, ten years after the formation of the league, with the first recipient being Bob Pettit. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) As of the 1980-1981 NBA season, the decision of which player receives the MVP award has been through voting by members of the media. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) This subjective evaluation means that there is not a concrete formula that determines the MVP award recipient. 

As a result of the uncertainty in MVP voting, researchers and analysts have tried to identify which statistical factors predict MVP outcomes. Research on the 2013-2023 MVP races used a Multiple Regression model with a mix of traditional statistics (points, total rebounds, assists, blocks, and steals) and advanced metrics (win shares, player efficiency rating, true shooting percentage, and usage rate) to examine their relationship with MVP voting outcomes. The results found that win shares, a player’s contribution to team success, had the greatest predictive strength of MVP rank <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Higher win share values were associated with a higher finish in MVP voting. This indicates modern MVP voting may be influenced by advanced metrics more than traditional box statistics.

A similar study by the Harvard Sports Analysis Collective also used a regression model to predict MVP outcomes. Focusing on the 1998-2018 MVP races, the model found that team wins was the most important predictor of a player’s MVP score <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). The final model predicted about 80% of the 1998-2018 MVPs correctly. Further analysis of the MVP winners that the model failed to predict helped reveal narrative as another factor of MVP voting. This suggests a potential roadblock in figuring out the player metrics that best predict MVP vote share because narratives are hard to quantify within statistical models.

Machine learning models have also been applied to predict MVP outcomes through player statistics. A study on predicting the 2018-2019 MVP on Kaggle implemented a Random Forest regression to estimate MVP award share based on a combination of traditional box score statistics and advanced metrics. Hyperparameter tuning with mean squared error indicated the Random Forest Regressor performed best with 500 estimators. <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) The Random Forest model correctly ranked the top 2 award share recipients for the 2018-2019 season, suggesting meaningful patterns in MVP voting can be captured with ensemble-based models. However, no additional seasons were evaluated, leaving room for analysis across different years.

Even though these studies were relatively successful, their scopes do not focus on the continuity or change of MVP voting over time. Since the introduction of advanced metrics in the 1996-1997 season, the relative importance of these statistics compared to traditional box score statistics has not been clearly established <a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6). Numerous sports analysts have debated which performance statistics best capture player value. A particularly influential work is Dean Oliver’s 2004 publication Basketball On Paper: Rules and Tools for Performance Analysis, in which he introduced the “Four Factors of Winning” as effective field goal percentage, offensive rebound rate, turnover percentage, and free throw rate <a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7). This suggests a transition toward efficiency rather than traditional counting stats, highlighting a potential turning point in how player performance is assessed for MVP voting.

## Hypothesis


- During 1996-2022, win shares will be the strongest MVP vote share predictor, followed by team wins and points per game.

  
The medium source’s regression result found that win shares had the most significant impact on MVP ranking. Moreover, their results showed that out of the five traditional box score stats (PTS, AST, TRB, STL, BLK), only points were statistically significant. In the Harvard model, team wins were the most significant, with the lowest p-value out of all tested predictors. 

- We predict advanced efficiency metrics (BPM, PER, TS%, etc) are stronger predictors of MVP vote share than traditional box score statistics (points, rebounds, assists, etc) in the modern era (2004-2022) than the pre-analytics era (1996-2004).
  
Although advanced metrics were introduced in 1996, we believe media voters would not readily adopt these new statistics in their MVP voting rationale. The introduction of Dean Oliver’s “Four Factors to Winning” likely changed the way advanced metrics were perceived, leading to greater credibility of advanced metrics for media voters. As a result, we predict that voters began to value the efficiency captured through advanced metrics more than the traditional counting statistics in the post-2004 era.
