## Research Question

Which factors predict NBA MVP voting share from 1996 to 2022, and how has their importance shifted across the eras?

- Which player statistics (points, assists, rebounds) and advanced efficiency metrics (BPM, Win Shares/48) are the strongest predictors of MVP vote share across the entire 1996-2022 period?
- How does the predictive importance of traditional (points, assists, rebounds) versus advanced metrics (BPM, Win Shares/48) differentiate between the pre-analytics era (1996-2004) and modern era (2004-2022)?

Our dataset combines per-game player statistics from Kaggle using manually added MVP vote sharing data and advanced stats from Basketball Reference. This means this is mainly a prediction + statistical inference task, and requires us to use regression-based models in order to count each statistic’s contribution to the vote share, and then compare the importance of features across both eras to identify whether the “MVP profile” changed alongside advanced analytics becoming mainstream.


## Background and Prior Work

The National Basketball Association (NBA) is the predominant professional basketball league in the United States. When people talk about the NBA, a common point of discussion is who the face of the league is. A tangible way to assign this title is by looking at which player is awarded the league MVP for each NBA season. The MVP award for the NBA was introduced in 1956, ten years after the formation of the league, with the first recipient being Bob Pettit. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) As of the 1980-1981 NBA season, the decision of which player receives the MVP award has been through voting by members of the media. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) This subjective evaluation means that there is not a concrete formula that determines the MVP award recipient. 

As a result of the uncertainty in MVP voting, researchers and analysts have tried to identify which statistical factors predict MVP outcomes. Coleman and his colleagues conducted early research on the MVP races determined by media voting, specifically the seasons from 1995-2005, to form models of media voting behavior. They used a mixture of player characteristics (e.g. race and basketball position), traditional box score statistics (e.g. points, total rebounds, and assists), and team metrics (e.g. team wins and playoff qualification) to build both a Tobit and a Probit Regression model of MVP voting <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Alongside addressing the original question on player race impacting MVP voting behavior, the models capture the significance of the traditional box score statistics. Both models indicated that a player’s points per game (PPG) was the most significant player statistic predictor of a player appearing on the MVP ballot (Probit) and the MVP voting points the player would receive (Tobit). However, the data used to build the models did not include any advanced metrics (e.g. BPM and win shares) which highlights potential confounding variables that their models did not capture.

More recent research by Li and Jia applies machine learning to identify the statistical features that best predict MVP outcomes. They utilized a dataset with both traditional and advanced metrics for NBA players from 1974 to 2024 to build the five machine learning models of Random Forest, LightGBM, XGBoost, CatBoost, and NGBoost. By considering a mixture of minimizing MSE and time consumption while maximizing the R-squared coefficient, CatBoost was found to be the optimal model <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). SHAP value analysis was used after further training of the CatBoost model to find that the most important predictors of the CatBoost’s player value scores are value over replacement player (VORP), points per game, and player efficiency rating (PER). These findings emphasize the importance of including advanced metrics in addition to traditional box score statistics in predicting MVP share. Yet, the treating of the data as a single time period leaves room for analysis across different eras and years.

Godbole and his colleagues explored MVP prediction further by training an ensemble of machine learning models across the three distinct NBA eras of the Ancient Era (1990-91 to 1999-00), the Medieval Era (2000-01 to 2011-12), and the Modern Era (2012-13 to 2021-22). They applied the five machine learning models of Support Vector Regression (SVR), ElasticNet, Adaptive Boosting, Random Forest, and Gradient Boosting as well as a backpropagation neural network that they named MVPNet. MVPNet outperformed all other models, achieving an R-squared coefficient of 0.86 and correctly predicting 10 out of 10 Modern Era MVPs <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5). SHAP analysis revealed that team statistics such as win percentage and seeding were the most important predictors in earlier eras, while advanced metrics grew more influential in the Modern Era. Although the study addresses how the predictive importance of statistics has shifted over time within each era, the primary focus of the study was on comparing how current players would perform in the MVP race across different eras.

While these studies deepen the understanding of MVP prediction, their scopes do not focus on the continuity or change of MVP voting over time. Since the complete tracking of advanced metrics in the 1996-1997 season, the relative importance of these statistics compared to traditional box score statistics has not been clearly established <a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6). Numerous sports analysts have debated which performance statistics best capture player value. A particularly influential work is Dean Oliver’s 2004 publication Basketball On Paper: Rules and Tools for Performance Analysis, in which he introduced the “Four Factors of Winning” as effective field goal percentage, offensive rebound rate, turnover percentage, and free throw rate <a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7). This suggests a transition toward efficiency rather than traditional counting stats, highlighting a potential turning point in how player performance is assessed for MVP voting. By using 2004 as a formal pivot point, we can test whether Oliver’s 'Four Factors' acted as a catalyst for media voters. While the 'Medieval' and 'Modern' eras are often grouped, our project specifically seeks to identify if 2004 was the precise moment the 'Points-Per-Game' dominance began to decay in favor of the efficiency-weighted profile Oliver championed. We are trying to capture how the predictive strength in advanced metrics compared to traditional box score statistics in gaining MVP vote share has changed over the time period spanning the 1996-1997 NBA season to the 2021-2022 NBA season.

## Hypothesis


- During 1996-2022, win shares per 48 minutes will be the strongest MVP vote share predictor, followed by points per game.

We expect win shares per 48 minutes to be the strongest MVP voter share predictor because it estimates the number of wins a player contributes to their team per 48 minutes, indicating whether a player’s performance is actually valuable to winning as opposed to just being high counting statistics with no real impact. We also expect points per game to be the next strongest predictor because high scoring output is the most visible aspect of a player’s game which would create a positive perception of the player by media voters.

- We predict advanced efficiency metrics (BPM, PER, TS%, etc) are stronger predictors of MVP vote share than traditional box score statistics (points, rebounds, assists, etc) in the modern era (2004-2022) than the pre-analytics era (1996-2004).
  
Although advanced metrics were introduced in 1996, we believe media voters would not readily adopt these new statistics in their MVP voting rationale. The introduction of Dean Oliver’s “Four Factors to Winning” likely changed the way advanced metrics were perceived, leading to greater credibility of advanced metrics for media voters. As a result, we predict that voters began to value the efficiency captured through advanced metrics more than the traditional counting statistics in the post-2004 era. This shift in voter focus was potentially reinforced by the 2004 elimination of hand-checking, which fundamentally changed scoring dynamics and made perimeter efficiency a more visible driver of team success.

### Analysis 1: Multiple Linear Regression on Full Time Period (1996-2022)

To investigate which statistical factors most strongly predict NBA MVP vote share, we fit a Multiple Linear Regression model on all player-seasons that received at least one MVP vote between 1996 and 2022. We created this subset to prevent the zero-inflated mvp vote share that Multiple Linear Regression does not handle well. The model regresses vote share against eleven predictors spanning both traditional box score metrics (points, assists, rebounds, steals, blocks, and turnovers) and advanced efficiency metrics (effective field goal percentage, usage rate, net rating, win shares per 48 minutes, and box plus-minus). All predictors are standardized prior to fitting so that the resulting coefficients are directly comparable. A larger absolute coefficient indicates a stronger influence on vote share regardless of the original unit of measurement. By examining the direction and magnitude of each standardized coefficient, this analysis allows us to quantify the relative importance of traditional versus advanced metrics in explaining MVP voting behavior across the full sample period.

```python
# This cell sets up and carries out the Multiple Linear Regression on players with non-zero MVP vote share across the entire period from (1996-2022)
# import OLS regression dependencies
from sklearn.preprocessing import StandardScaler
import statsmodels.formula.api as smf

traditional_vars = ['pts', 'ast', 'reb', 'STL', 'BLK', 'TOV']
advanced_vars = ['eFG%', 'usg_pct', 'net_rating', 'WS/48', 'BPM', 'TOV%']
all_predictors = traditional_vars + advanced_vars
target = 'share'

# subset to vote-getters only
df_mvp = df_merged[df_merged['share'] > 0].copy()

# standardize predictors so coefficients are directly comparable
scaler = StandardScaler()
df_mvp[all_predictors] = scaler.fit_transform(df_mvp[all_predictors])

# build formula string e.g. 'share ~ pts + ast + reb + ...'
# rename columns with '/' or '%' to make a clean formula
df_mvp = df_mvp.rename(columns={
    'WS/48': 'WS_per_48',
    'eFG%':  'eFG_pct',
    'TOV%':  'TOV_pct'
})
all_predictors_clean = ['pts', 'ast', 'reb', 'STL', 'BLK', 'TOV',
                        'eFG_pct', 'usg_pct', 'net_rating', 'WS_per_48', 'BPM', 'TOV_pct']

formula = 'share ~ ' + ' + '.join(all_predictors_clean)
print("Using the Formula:", formula)

# fit model
print("\n========== FULL-PERIOD MLR MODEL (1996–2022) ==========")
model_full = smf.ols(formula, data=df_mvp).fit()
print(model_full.summary())

# plot coefficients
coef_df = pd.DataFrame({
    'variable': all_predictors_clean,
    'coef':     [model_full.params[v] for v in all_predictors_clean],
    'ci_low':   [model_full.conf_int().loc[v, 0] for v in all_predictors_clean],
    'ci_high':  [model_full.conf_int().loc[v, 1] for v in all_predictors_clean],
    'pval':     [model_full.pvalues[v] for v in all_predictors_clean]
}).sort_values('coef', ascending=True)

fig, ax = plt.subplots(figsize=(9, 7))
colors = ['#2196F3' if c > 0 else '#FF5722' for c in coef_df['coef']]
ax.barh(coef_df['variable'], coef_df['coef'], color=colors, alpha=0.85, edgecolor='white')
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')
ax.set_xlabel('Standardized Coefficient (β)', fontsize=11)
ax.set_ylabel('Feature', fontsize=11)
ax.set_title('Full-Period MLR: Standardized Coefficients (1996–2022)', fontsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

The full-period OLS regression model explains approximately 47% of the variance in MVP vote share (R² = 0.470, Adjusted R² = 0.453), suggesting that player statistics capture a meaningful but incomplete picture of MVP voting behavior. Among individual predictors, net rating (β = 0.046, p = 0.002) and WS/48 (β = 0.105, p < 0.001) emerged as the only statistically significant positive predictors, suggesting that overall team impact and efficiency are the strongest measurable drivers of MVP recognition. Another notable outcome is that traditional box score statistics including points, assists, rebounds, steals, and blocks all had relatively low coefficient absolute values and a higher p-value (p > 0.05). Effective field goal percentage was the only significant negative predictor (β = -0.040, p = 0.019), which seems counterintuitive, but may be a result of MVP vote share being more guard dominated during this time period as centers generally have higher eFG%. Several important limitations apply to these results. The non-normal residuals confirmed by the Omnibus test (p = 0.000) and Jarque-Bera test (p = 4.08e-11) mean that reported p-values and confidence intervals are approximations. Additionally, by restricting the sample to players who already received MVP votes, the model captures what differentiates MVP candidates from each other but cannot be generalized to explain what determines MVP candidacy in the first place. These limitations collectively motivate the Random Forest analysis that follows, which makes no distributional assumptions and handles correlated predictors more robustly through its ensemble structure.

### Analysis 2: Multiple Linear Regression Stratified by Era

To investigate whether the statistical predictors of NBA MVP vote share shifted across eras, we stratify the same vote-getter subset into two periods defined by the 2004 publication of Dean Oliver's Basketball on Paper: the Pre-Analytics Era (1996–2004) and the Modern Era (2004–2022). A separate Multiple Linear Regression model is fit on each era using the same eleven predictors as the full-period model, with predictors re-standardized within each era so that coefficients reflect the relative importance of each stat in that specific period rather than across the full sample. By comparing the standardized coefficients side by side, this analysis allows us to directly test our hypothesis of advanced efficiency metrics gaining predictive power over traditional box score statistics in the modern era as media voters became more exposed to analytics-based evaluation.

```python
# This cell sets up and carries out the Multiple Linear Regression on players with non-zero MVP vote share stratified by
# the pre-analytics (1996-1997 to 2003-2004) and the modern eras (2004-2005 to 2021-2022)

# add era column to df_mvp
df_mvp['start_year'] = df_mvp['season'].str[:4].astype(int)
df_mvp['era'] = df_mvp['start_year'].apply(
    lambda s: 'pre_analytics' if s < 2004 else 'modern'
)

era_models = {}

# fit model for each era
for era_name, label in [('pre_analytics', 'Pre-Analytics Era (1996–2004)'),
                         ('modern', 'Modern Era (2004–2022)')]:
    df_era = df_mvp[df_mvp['era'] == era_name].copy()

    # re-standardize within era
    scaler = StandardScaler()
    df_era[all_predictors_clean] = scaler.fit_transform(df_era[all_predictors_clean])

    model_era = smf.ols(formula, data=df_era).fit()
    era_models[era_name] = model_era

    print(f"\n========== {label} ==========")
    print(model_era.summary())

# plot coefficients for each era in a side by side bar graph
coef_rows = []
for era_name, label in [('pre_analytics', 'Pre-Analytics (1996-2004)'),
                         ('modern', 'Modern (2004-2022)')]:
    m = era_models[era_name]
    for var in all_predictors_clean:
        coef_rows.append({
            'era': label,
            'variable': var,
            'coef': m.params[var]
        })

coef_df_era = pd.DataFrame(coef_rows)
fig, ax = plt.subplots(figsize=(11, 7))
x = np.arange(len(all_predictors_clean))
width = 0.35
colors = {
    'Pre-Analytics (1996-2004)': '#2196F3',
    'Modern (2004-2022)': '#FF5722'
}
for i, (era, color) in enumerate([('Pre-Analytics (1996-2004)', '#2196F3'),
                                   ('Modern (2004-2022)', '#FF5722')]):
    group = coef_df_era[coef_df_era['era'] == era]
    offset = (i - 0.5) * width

    ax.bar(x + offset, group['coef'], width,
           label=era, color=color, alpha=0.85, edgecolor='white')

ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.set_xticks(x)
ax.set_xticklabels(all_predictors_clean, rotation=35, ha='right', fontsize=10)
ax.set_ylabel('Standardized Coefficient (β)', fontsize=11)
ax.set_title('Era Comparison: Standardized Regression Coefficients\n'
             'Pre-Analytics (1996-2004) vs. Modern (2004-2022)', fontsize=12)
ax.legend(fontsize=10)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

The era-stratified models reveal meaningful shifts in how statistical factors predicted MVP vote share across the two periods. In the Pre-Analytics Era (1996–2004), the model explains 48% of the variance (R² = 0.480, Adjusted R² = 0.431) with points per game emerging as strongest positive predictor (β = 0.178, p = 0.025), consistent with the expectation that traditional scoring output was the dominant driver of MVP recognition during this period. Effective field goal percentage was the strongest  negative predictor (β = -0.079, p = 0.007), echoing the full-period finding. In the Modern Era (2004–2022), the model explains a slightly higher 51% of variance (R² = 0.511, Adjusted R² = 0.486). The predictor landscape shifts notably as points per game shows negligible predictive power (β = -0.001, p = 0.982), while WS/48 emerges as the strongest positive predictor (β = 0.129, p < 0.001) and BPM shifts from a negative predictor to a positive predictor (β = -0.010 to β = 0.057). This shift from points to win shares as the dominant predictor across eras provides direct support for our hypothesis that advanced efficiency metrics gained predictive power in the modern era as media voters increasingly incorporated analytics-based evaluation into their decision making. However, both models share the same limitations as the full-period model with non-normal residuals and a restricted sample of only MVP vote-getters.

### Analysis 3: Ridge Regression Stratified by Era (Pre- vs Post-2004)

Analysis 1 fit a single model across the full period. To test that the predictive importance of advanced metrics shifted after the 2003-04 rule, we use **Ridge regression** rather than ordinary least squares because our predictors are highly collinear (see the correlation heatmap above): metrics such as PER, BPM, WS/48, and net rating have overlapping information, making plain OLS coefficients unstable and difficult to interpret. Ridge adds an L2 penalty that shrinks coefficients toward zero, which stabilizes them in the presence of collinearity so that we reduce the noise and the comparison shows real signals. 

```python
from sklearn.linear_model import RidgeCV
from sklearn.preprocessing import StandardScaler

traditional_vars = ['pts', 'ast', 'reb', 'STL', 'BLK', 'TOV']
advanced_vars    = ['eFG%', 'usg_pct', 'net_rating', 'WS/48', 'BPM', 'TOV%']
all_predictors   = traditional_vars + advanced_vars
target           = 'share'

# Vote-getters only 
df_era = df_merged[df_merged['share'] > 0].copy()

df_era['start_year'] = df_era['season'].str[:4].astype(int)
df_era['era'] = np.where(df_era['start_year'] >= 2004, 'Modern (2004-2022)', 'Pre-Analytics (1996-2004)')

# Helper function: fit a RidgeCV model on one era's data. Reusable for all eras
def fit_ridge_for_era(era_df, predictors, target):
    scaler = StandardScaler()
    X = scaler.fit_transform(era_df[predictors])
    y = era_df[target].values

    # RidgeCV tries several penalty strengths and picks the best one through cross-validation
    alphas = np.logspace(-3, 3, 50)        # 50 candidates for penalty strengths
    model = RidgeCV(alphas=alphas).fit(X, y) # Tries all 50 candidates and fit the best one

    coefs = pd.Series(model.coef_, index=predictors)
    r2 = model.score(X, y)              
    return model, coefs, r2, model.alpha_

# Fit one model per era
results = {}
for era_name, era_df in df_era.groupby('era'):
    model, coefs, r2, alpha = fit_ridge_for_era(era_df, all_predictors, target)
    results[era_name] = {'coefs': coefs, 'r2': r2, 'alpha': alpha, 'n': len(era_df)}
    print(f"{era_name}: n={len(era_df)}, R^2={r2:.3f}, chosen alpha={alpha:.3g}")

# coefficient table (pre vs post)
pre_name  = 'Pre-Analytics (1996-2004)'
post_name = 'Modern (2004-2022)'
coef_compare = pd.DataFrame({
    'Pre-2004':  results[pre_name]['coefs'],
    'Post-2004': results[post_name]['coefs'],
})

coef_compare['change'] = coef_compare['Post-2004'] - coef_compare['Pre-2004']
coef_compare['type'] = ['Traditional' if v in traditional_vars else 'Advanced' for v in coef_compare.index]

print("\nStandardized Ridge coefficients by era:")
print(coef_compare.round(4))

order = coef_compare.sort_values('Post-2004').index
y_pos = np.arange(len(order))
bar_h = 0.4

fig, ax = plt.subplots(figsize=(9, 8))
ax.barh(y_pos + bar_h/2, coef_compare.loc[order, 'Pre-2004'],
        height=bar_h, label='Pre-2004', color='#5B9BD5', alpha=0.9, edgecolor='white')
ax.barh(y_pos - bar_h/2, coef_compare.loc[order, 'Post-2004'],
        height=bar_h, label='Post-2004', color='#ED7D31', alpha=0.9, edgecolor='white')
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')
ax.set_yticks(y_pos)
ax.set_yticklabels(order)
ax.set_xlabel('Standardized Ridge Coefficient (β)', fontsize=11)
ax.set_ylabel('Feature', fontsize=11)
ax.set_title('Ridge Regression by Era: Standardized Coefficients (Pre vs Post 2004)', fontsize=12)
ax.legend()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```
